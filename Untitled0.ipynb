## 4. TRAIN / VALIDATION SPLIT

To prevent data leakage, we will split the data based on `source1_entity_id`s. This means that all entries related to a specific `source1_entity_id` will either be entirely in the training set or entirely in the validation set. We will focus on splitting the `df_gt` (ground truth) first, as it links `source1_entity_id`s to their matches.

## 5. TEXT PREPROCESSING / NORMALIZATION

This section focuses on cleaning and standardizing the text data in the `business_name`, `business_address`, and `country` columns. These steps are crucial for improving the quality of the matching process, as variations in casing, punctuation, or formatting can lead to missed matches. We will apply these transformations to all training and validation DataFrames (`df_s1_train`, `df_s1_val`, `df_s2_train`, `df_s2_val`, `df_s3_train`, `df_s3_val`).

```python
import re
import string

def clean_text(text):
    if pd.isna(text) or text is None:
        return text
    # Convert to lowercase
    text = str(text).lower()
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Normalize whitespace (replace multiple spaces with a single space, strip leading/trailing spaces)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

def normalize_country(country):
    if pd.isna(country) or country is None:
        return country
    return str(country).lower()

# List of DataFrames to preprocess
dfs_to_process = [
    df_s1_train, df_s1_val,
    df_s2_train, df_s2_val,
    df_s3_train, df_s3_val
]

# Apply cleaning and normalization
for df in dfs_to_process:
    for col in ['business_name', 'business_address']:
        df[col] = df[col].apply(clean_text)
    df['country'] = df['country'].apply(normalize_country)

print("Text preprocessing (lowercase, punctuation removal, whitespace normalization) and country normalization applied to all dataframes.")

# Display head of processed dataframes to verify
print("\n--- df_s1_train after preprocessing ---")
display(df_s1_train.head())

print("\n--- df_s2_train after preprocessing ---")
display(df_s2_train.head())

print("\n--- df_s3_train after preprocessing ---")
display(df_s3_train.head())
```

### 5.1 Further Text Normalization: Abbreviations and Suffixes

To further standardize the text data, we will expand common abbreviations in `business_address` and normalize various business suffixes in `business_name`. This helps in reducing variations and improving the accuracy of later matching processes. We will apply these transformations to all training and validation DataFrames.

```python
import re

# Define common abbreviations and their expansions for addresses
abbreviation_map = {
    'st': 'street',
    'ave': 'avenue',
    'rd': 'road',
    'dr': 'drive',
    'ln': 'lane',
    'ct': 'court',
    'pl': 'place',
    'sq': 'square',
    'blvd': 'boulevard',
    'hwy': 'highway',
    'pkwy': 'parkway',
    'apt': 'apartment',
    'ste': 'suite',
    'po box': 'post office box',
    'rte': 'route',
    'cir': 'circle',
    'fwy': 'freeway',
    'jct': 'junction',
    'mt': 'mount',
    'n': 'north',
    's': 'south',
    'e': 'east',
    'w': 'west',
    'ne': 'northeast',
    'nw': 'northwest',
    'se': 'southeast',
    'sw': 'southwest',
    'bldg': 'building',
    'dept': 'department',
    'fl': 'floor',
    'rm': 'room',
    'unit': 'unit'
}

# Define common business suffixes to normalize or remove
suffix_map = {
    'inc': 'incorporated',
    'llc': 'limited liability company',
    'ltd': 'limited',
    'corp': 'corporation',
    'co': 'company',
    'bros': 'brothers',
    'prvt ltd': 'private limited',
    'pvt ltd': 'private limited',
    'pvt': 'private',
    'lp': 'limited partnership',
    'llp': 'limited liability partnership',
    'pc': 'professional corporation',
    'pllc': 'professional limited liability company'
}

def expand_abbreviations(text, abbr_map):
    if pd.isna(text) or text is None:
        return text
    words = text.split()
    expanded_words = []
    for word in words:
        lower_word = word.lower()
        if lower_word in abbr_map:
            expanded_words.append(abbr_map[lower_word])
        else:
            expanded_words.append(word)
    return ' '.join(expanded_words)

def normalize_suffixes(text, suffix_map):
    if pd.isna(text) or text is None:
        return text
    # Sort suffixes by length descending to handle 'prvt ltd' before 'ltd'
    # Iterate over items (key, value) instead of just keys
    sorted_suffixes_items = sorted(suffix_map.items(), key=lambda item: len(item[0]), reverse=True)

    for suffix_abbr, suffix_full in sorted_suffixes_items:
        # Use word boundaries to avoid replacing parts of other words
        text = re.sub(r'\b' + re.escape(suffix_abbr) + r'\b', suffix_full, text)

    # Remove duplicate spaces created by replacements
    text = re.sub(r'\s+', ' ', text).strip()
    return text


# Apply expansion and normalization
for df in dfs_to_process:
    # Apply abbreviation expansion to business_address
    df['business_address'] = df['business_address'].apply(lambda x: expand_abbreviations(x, abbreviation_map))

    # Apply suffix normalization to business_name
    df['business_name'] = df['business_name'].apply(lambda x: normalize_suffixes(x, suffix_map))

print("Abbreviation expansion and suffix normalization applied to all dataframes.")

# Display head of processed dataframes to verify
print("\n--- df_s1_train after abbreviation/suffix normalization ---")
display(df_s1_train.head())

print("\n--- df_s2_train after abbreviation/suffix normalization ---")
display(df_s2_train.head())

print("\n--- df_s3_train after abbreviation/suffix normalization ---")
display(df_s3_train.head())
```

### 5.2 Number and Postal Code Normalization for Addresses & Tokenization for Names

To further clean and standardize the `business_address` and `business_name` fields, we will implement more specific normalization:

*   **`business_address`**: Remove or standardize numerical components (like house numbers, apartment numbers) and postal codes. For this stage, we will aim to remove common numerical patterns that might not be useful for matching or could introduce noise. We will also remove postal codes as they are often very specific and might hinder matching across slight variations.
*   **`business_name`**: Further tokenize and clean names to handle multiple words and ensure consistency. This often involves removing common stop words or very short, non-descriptive tokens, though for initial matching, keeping most tokens is often better. For now, we will focus on more general token cleanup.


```python
import re
import pandas as pd
import gc

us_zip_pattern = re.compile(r'\b\d{5}(?:-\d{4})?\b')
postal_pattern = re.compile(r'\b\d{6}\b')
number_pattern = re.compile(r'\b\d+\b')
space_pattern = re.compile(r'\s+')

def normalize_address(text):
    if pd.isna(text):
        return text

    text = str(text)
    text = us_zip_pattern.sub('', text)
    text = postal_pattern.sub('', text)
    text = number_pattern.sub('', text)
    text = space_pattern.sub(' ', text).strip()

    return text

def normalize_name(text):
    if pd.isna(text):
        return text

    return space_pattern.sub(' ', str(text)).strip()

dfs_to_process = [
    df_s1_train,
    df_s1_val,
    df_s2_train,
    df_s2_val,
    df_s3_train,
    df_s3_val
]

for df in dfs_to_process:
    df['business_address'] = df['business_address'].map(
        normalize_address
    )

    df['business_name'] = df['business_name'].map(
        normalize_name
    )

del dfs_to_process
gc.collect()

print("Address and business-name normalization completed.")

print("\nS1 Train:")
display(df_s1_train.head())

print("\nS2 Train:")
display(df_s2_train.head())

print("\nS3 Train:")
display(df_s3_train.head())
```

## 6. CANDIDATE GENERATION / BLOCKING

Candidate generation, also known as blocking, is a critical step in entity resolution. Its purpose is to significantly reduce the number of potential pairs that need to be compared, making the matching process computationally feasible. Instead of comparing every record from one source with every record from another (a Cartesian product), blocking strategies create smaller, more manageable subsets of records that are likely to contain matches. The goal here is to achieve high recall, ensuring that most true matches are included in the candidate set.

### 6.1 Exact Name Blocking

This strategy involves pairing records that have identical `business_name` values after preprocessing. This is a simple yet effective blocking method for capturing highly similar entities.

```python
import pandas as pd
import gc

def exact_name_blocking(df1, df2):
    left = df1[
        ['entity_id', 'business_name']
    ].dropna(subset=['business_name'])

    right = df2[
        ['entity_id', 'business_name']
    ].dropna(subset=['business_name'])

    left = left.drop_duplicates(
        subset=['entity_id', 'business_name']
    )

    right = right.drop_duplicates(
        subset=['entity_id', 'business_name']
    )

    result = pd.merge(
        left,
        right,
        on='business_name',
        how='inner',
        sort=False
    )

    result = result[
        ['entity_id_x', 'entity_id_y']
    ]

    result.columns = [
        'entity_id_s1',
        'entity_id_s2_or_s3'
    ]

    result = result.drop_duplicates(
        ignore_index=True
    )

    del left
    del right
    gc.collect()

    return result


print("Generating exact-name candidates...")


candidates_s1_s2_train_exact_name = exact_name_blocking(
    df_s1_train,
    df_s2_train
)

print(
    "S1-S2 Train:",
    candidates_s1_s2_train_exact_name.shape
)

gc.collect()


candidates_s1_s3_train_exact_name = exact_name_blocking(
    df_s1_train,
    df_s3_train
)

print(
    "S1-S3 Train:",
    candidates_s1_s3_train_exact_name.shape
)

gc.collect()


candidates_s1_s2_val_exact_name = exact_name_blocking(
    df_s1_val,
    df_s2_val
)

print(
    "S1-S2 Validation:",
    candidates_s1_s2_val_exact_name.shape
)

gc.collect()


candidates_s1_s3_val_exact_name = exact_name_blocking(
    df_s1_val,
    df_s3_val
)

print(
    "S1-S3 Validation:",
    candidates_s1_s3_val_exact_name.shape
)

gc.collect()


print("\nExact-name blocking completed.")
```

### 6.2 Token Blocking

Token blocking is a more flexible blocking strategy than exact name blocking. It involves generating candidate pairs based on shared tokens (words) within the `business_name`. This can help capture matches where names have slight variations, word order changes, or extra descriptive words. We will create blocks based on individual words in the `business_name`.

```python
from collections import defaultdict
import pandas as pd
import gc

def create_rare_token_blocks(df, name_col='business_name', id_col='entity_id', max_entities=1000):
    blocks = defaultdict(list)

    data = df[[name_col, id_col]].dropna(
        subset=[name_col, id_col]
    )

    for name, entity_id in data.itertuples(index=False, name=None):
        tokens = set(str(name).lower().split())

        for token in tokens:
            if len(token) > 2:
                blocks[token].append(entity_id)

    blocks = {
        token: ids
        for token, ids in blocks.items()
        if len(ids) <= max_entities
    }

    del data
    gc.collect()

    return blocks


def generate_token_candidates_safe(
    blocks_s1,
    blocks_sx,
    max_pairs=2000000
):
    candidates = set()

    common_tokens = blocks_s1.keys() & blocks_sx.keys()

    for token in common_tokens:
        ids1 = blocks_s1[token]
        idsx = blocks_sx[token]

        if len(candidates) + len(ids1) * len(idsx) > max_pairs:
            continue

        for id1 in ids1:
            for idx in idsx:
                candidates.add((id1, idx))

    result = pd.DataFrame(
        list(candidates),
        columns=['entity_id_s1', 'entity_id_s2_or_s3']
    )

    del candidates
    del common_tokens
    gc.collect()

    return result
```

```python
print("Creating S1-S2 training token blocks...")

blocks_s1 = create_rare_token_blocks(df_s1_train)
blocks_s2 = create_rare_token_blocks(df_s2_train)

print("S1 tokens:", len(blocks_s1))
print("S2 tokens:", len(blocks_s2))

candidates_s1_s2_train_token_block = generate_token_candidates_safe(
    blocks_s1,
    blocks_s2,
    max_pairs=2000000
)

print(
    "S1-S2 Train candidates:",
    candidates_s1_s2_train_token_block.shape
)

del blocks_s1
del blocks_s2
gc.collect()
```

```python
import pandas as pd
import numpy as np
import gc

def build_features_chunked(
    candidates,
    df_s1,
    df_s2,
    chunk_size=100000
):
    s1 = df_s1[
        ['entity_id', 'business_name', 'business_address']
    ].drop_duplicates('entity_id').set_index('entity_id')

    s2 = df_s2[
        ['entity_id', 'business_name', 'business_address']
    ].drop_duplicates('entity_id').set_index('entity_id')

    feature_chunks = []

    for start in range(0, len(candidates), chunk_size):

        chunk = candidates.iloc[start:start + chunk_size]

        left = s1.reindex(
            chunk['entity_id_s1'].values
        ).reset_index()

        right = s2.reindex(
            chunk['entity_id_s2_or_s3'].values
        ).reset_index()

        name1 = left['business_name'].astype('string').fillna('').str.lower()
        name2 = right['business_name'].astype('string').fillna('').str.lower()

        addr1 = left['business_address'].astype('string').fillna('').str.lower()
        addr2 = right['business_address'].astype('string').fillna('').str.lower()

        name1_values = name1.tolist()
        name2_values = name2.tolist()
        addr1_values = addr1.tolist()
        addr2_values = addr2.tolist()

        name_exact = []
        name_jaccard = []
        name_overlap = []

        address_exact = []
        address_jaccard = []
        address_overlap = []

        for a, b in zip(name1_values, name2_values):
            set_a = set(a.split())
            set_b = set(b.split())
            union = set_a | set_b

            name_exact.append(int(a == b))
            name_overlap.append(len(set_a & set_b))
            name_jaccard.append(
                len(set_a & set_b) / len(union)
                if union else 0.0
            )

        for a, b in zip(addr1_values, addr2_values):
            set_a = set(a.split())
            set_b = set(b.split())
            union = set_a | set_b

            address_exact.append(int(a == b))
            address_overlap.append(len(set_a & set_b))
            address_jaccard.append(
                len(set_a & set_b) / len(union)
                if union else 0.0
            )

        features = pd.DataFrame({
            'name_exact': np.asarray(name_exact, dtype=np.int8),
            'name_jaccard': np.asarray(name_jaccard, dtype=np.float32),
            'name_token_overlap': np.asarray(name_overlap, dtype=np.int8),
            'address_exact': np.asarray(address_exact, dtype=np.int8),
            'address_jaccard': np.asarray(address_jaccard, dtype=np.float32),
            'address_token_overlap': np.asarray(address_overlap, dtype=np.int16)
        })

        feature_chunks.append(features)

        del chunk
        del left
        del right
        del name1
        del name2
        del addr1
        del addr2
        del name1_values
        del name2_values
        del addr1_values
        del addr2_values
        del name_exact
        del name_jaccard
        del name_overlap
        del address_exact
        del address_jaccard
        del address_overlap
        del features

        gc.collect()

        print(
            f"Processed {min(start + chunk_size, len(candidates)):,} "
            f"/ {len(candidates):,}"
        )

    result = pd.concat(
        feature_chunks,
        ignore_index=True
    )

    del feature_chunks
    del s1
    del s2

    gc.collect()

    return result
```

```python
import pandas as pd
import gc
from collections import defaultdict

def create_high_recall_blocks(
    df,
    name_col='business_name',
    address_col='business_address',
    id_col='entity_id'
):
    blocks = defaultdict(list)

    data = df[
        [id_col, name_col, address_col]
    ].dropna(subset=[id_col])

    for entity_id, name, address in data.itertuples(
        index=False,
        name=None
    ):
        name = '' if pd.isna(name) else str(name).lower()
        address = '' if pd.isna(address) else str(address).lower()

        name_tokens = set(name.split())
        address_tokens = set(address.split())

        # 1. Exact normalized name
        if name:
            blocks[f"name_exact:{name}"].append(entity_id)

            # 2. Name prefix
            if len(name) >= 4:
                blocks[
                    f"name_prefix:{name[:4]}"
                ].append(entity_id)

        # 3. Name tokens
        for token in name_tokens:
            if len(token) >= 3:
                blocks[
                    f"name_token:{token}"
                ].append(entity_id)

        # 4. Address tokens
        for token in address_tokens:
            if len(token) >= 4:
                blocks[
                    f"addr_token:{token}"
                ].append(entity_id)

    del data
    gc.collect()

    return blocks
```

```python
import pandas as pd
import gc

def generate_high_recall_candidates(
    blocks_left,
    blocks_right,
    max_entities_per_block=200,
    max_candidates=2_000_000
):
    candidates = set()

    common_keys = blocks_left.keys() & blocks_right.keys()

    for key in common_keys:

        left_ids = blocks_left[key]
        right_ids = blocks_right[key]

        # Skip extremely common blocks
        if (
            len(left_ids) > max_entities_per_block
            or len(right_ids) > max_entities_per_block
        ):
            continue

        pair_count = len(left_ids) * len(right_ids)

        # Stop before candidate set becomes too large
        if len(candidates) + pair_count > max_candidates:
            continue

        for left_id in left_ids:
            for right_id in right_ids:
                candidates.add((left_id, right_id))

    result = pd.DataFrame(
        list(candidates),
        columns=[
            'entity_id_s1',
            'entity_id_s2_or_s3'
        ]
    )

    del candidates
    del common_keys
    gc.collect()

    return result
```

```python
import re
import pandas as pd
import gc

def name_fingerprint(value):
    if pd.isna(value):
        return ""

    value = str(value).lower()

    # Remove punctuation
    value = re.sub(r'[^a-z0-9]', '', value)

    # Remove common business suffixes
    suffixes = [
        'incorporated',
        'corporation',
        'company',
        'limited',
        'llc',
        'ltd',
        'corp',
        'inc'
    ]

    for suffix in suffixes:
        if value.endswith(suffix) and len(value) > len(suffix) + 3:
            value = value[:-len(suffix)]
            break

    return value[:8]
```

```python
def create_fingerprint_blocks(df):
    blocks = {}

    temp = df[
        ['entity_id', 'business_name']
    ].dropna(subset=['entity_id']).copy()

    temp['name_fp'] = temp['business_name'].map(
        name_fingerprint
    )

    temp = temp[
        temp['name_fp'].str.len() >= 4
    ]

    for fp, group in temp.groupby('name_fp', sort=False):
        blocks[fp] = group['entity_id'].tolist()

    del temp
    gc.collect()

    return blocks
```

```python
def generate_fingerprint_candidates(
    blocks_left,
    blocks_right,
    max_entities_per_block=100
):
    candidates = set()

    common_keys = blocks_left.keys() & blocks_right.keys()

    for key in common_keys:
        left_ids = blocks_left[key]
        right_ids = blocks_right[key]

        # Skip very common fingerprints
        if (
            len(left_ids) > max_entities_per_block
            or len(right_ids) > max_entities_per_block
        ):
            continue

        for left_id in left_ids:
            for right_id in right_ids:
                candidates.add((left_id, right_id))

    result = pd.DataFrame(
        list(candidates),
        columns=[
            'entity_id_s1',
            'entity_id_s2_or_s3'
        ]
    )

    del candidates
    del common_keys
    gc.collect()

    return result
```

```python
# ============================================
# FINGERPRINT BLOCKING RECALL
# ============================================

def calculate_recall_merge(candidates, gt_pairs):
    gt_df = pd.DataFrame(
        list(gt_pairs),
        columns=['entity_id_s1', 'entity_id_s2_or_s3']
    )

    found = candidates.merge(
        gt_df,
        on=['entity_id_s1', 'entity_id_s2_or_s3'],
        how='inner'
    )

    recall = len(found) / len(gt_pairs) if gt_pairs else 0

    del gt_df
    del found
    gc.collect()

    return recall


fp_recall_s2 = calculate_recall_merge(
    candidates_fp_s2,
    gt_val_s2
)

fp_recall_s3 = calculate_recall_merge(
    candidates_fp_s3,
    gt_val_s3
)

print(f"S1-S2 fingerprint recall: {fp_recall_s2 * 100:.4f}%")
print(f"S1-S3 fingerprint recall: {fp_recall_s3 * 100:.4f}%")
```

```python
import re
from collections import defaultdict
import gc

def address_fingerprint(value):
    if pd.isna(value):
        return ""

    value = str(value).lower()

    # Keep alphanumeric characters
    value = re.sub(r'[^a-z0-9]', '', value)

    # Remove common address words/suffixes
    replacements = [
        'street',
        'road',
        'avenue',
        'boulevard',
        'drive',
        'lane',
        'highway',
        'building',
        'apartment',
        'suite',
        'floor'
    ]

    for word in replacements:
        value = value.replace(word, '')

    return value[:10]


def create_address_fingerprint_blocks(df):
    blocks = {}

    temp = df[
        ['entity_id', 'business_address']
    ].dropna(
        subset=['entity_id']
    ).copy()

    temp['addr_fp'] = temp['business_address'].map(
        address_fingerprint
    )

    temp = temp[
        temp['addr_fp'].str.len() >= 5
    ]

    for fp, group in temp.groupby(
        'addr_fp',
        sort=False
    ):
        blocks[fp] = group['entity_id'].tolist()

    del temp
    gc.collect()

    return blocks


print("Creating address fingerprint blocks...")

addr_fp_s1 = create_address_fingerprint_blocks(df_s1_val)
addr_fp_s2 = create_address_fingerprint_blocks(df_s2_val)
addr_fp_s3 = create_address_fingerprint_blocks(df_s3_val)

print("S1 address blocks:", f"{len(addr_fp_s1):,}")
print("S2 address blocks:", f"{len(addr_fp_s2):,}")
print("S3 address blocks:", f"{len(addr_fp_s3):,}")
```
